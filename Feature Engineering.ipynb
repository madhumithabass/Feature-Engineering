## **Feature Engineering**

Whatever we do with features, we call it as Feature Engineering.

* **Feature Elimination** – dropping the features.
* **Feature Addition** – adding some features.
* **Feature Transformation** – transforming the given feature values into another scale - Log Transformation, Sqrt Transformation.
* **Feature Selection** – deciding which features are important out of many features and choosing that features for model building.

## **Feature Selection Techniques**

* sklearn - SelectFromModel


* sklearn - RFE(ie, Recursive Feature Elimination)

## Import Necesasary Libraries

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import RFE,SelectFromModel
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score
from sklearn.ensemble import RandomForestClassifier

import warnings
warnings.filterwarnings("ignore")
```

## Data Collection, Understanding and Preparation

```python
breast_data = load_breast_cancer()

breast_data_df = pd.DataFrame(breast_data["data"], columns=breast_data["feature_names"])
breast_data_df["Target"] = breast_data["target"]

## Split the dataset into X and y

X = breast_data_df.drop("Target", axis = 1)
y = breast_data_df["Target"]

## Split this into training and testing data

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,shuffle=True,stratify=y)

print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

## Model Building, Training, Testing and Model Evaluation

```python
## Model Building

RFE = RandomForestClassifier(max_depth=7)

## Model Training

RFE_classifier = RFE.fit(X_train,y_train)

## Model Testing

y_pred_train = RFE_classifier.predict(X_train)
y_pred_test = RFE_classifier.predict(X_test)

## Model Evaluation

print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
print("Accuracy Score:\t", accuracy_score(y_test,y_pred_test))
print("Precision Score:\t", precision_score(y_test,y_pred_test))
print("Recall Score:\t", recall_score(y_test,y_pred_test))
```

## Let's use Feature Selection

## 1. Select From Model - Feature Selection

```python
Select_from_model = SelectFromModel(estimator = RandomForestClassifier(), max_features=None)

Select_from_model.fit(X_train,y_train)
```

```python
## Gives the selected features in boolean

Select_from_model.get_support()
```

```python
## The feature name

breast_data.feature_names 
```

```python
print(breast_data.feature_names[Select_from_model.get_support()])
print("\nTotal Selected Features:", len(breast_data.feature_names[Select_from_model.get_support()]))
```

```python
X_train_select_from_model = Select_from_model.transform(X_train)
X_test_select_from_model = Select_from_model.transform(X_test)
```

```python
## Model Building

RFE = RandomForestClassifier(max_depth=7)

## Model Training

RFE_classifier = RFE.fit(X_train_select_from_model,y_train)

## Model Testing

y_pred_train = RFE_classifier.predict(X_train_select_from_model)
y_pred_test = RFE_classifier.predict(X_test_select_from_model)

## Model Evaluation

print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
print("Accuracy Score:\t", accuracy_score(y_test,y_pred_test))
print("Precision Score:\t", precision_score(y_test,y_pred_test))
print("Recall Score:\t", recall_score(y_test,y_pred_test))
```

## Create a User defined Function

```python
def runRandomForestClassifier(X_train,X_test,y_train,y_test):

    ## Model Building
    
    RFE = RandomForestClassifier(max_depth=7, random_state=123)
    
    ## Model Training
    
    RFE_classifier = RFE.fit(X_train,y_train)
    
    ## Model Testing
    
    y_pred_train = RFE_classifier.predict(X_train)
    y_pred_test = RFE_classifier.predict(X_test)
    
    ## Model Evaluation
    
    print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
    print("Accuracy Score:\t", accuracy_score(y_test,y_pred_test))
    print("Precision Score:\t", precision_score(y_test,y_pred_test))
    print("Recall Score:\t", recall_score(y_test,y_pred_test))
```

```python
%%time

runRandomForestClassifier(X_train,X_test,y_train,y_test)  ## 30 features
```

```python
%%time

runRandomForestClassifier(X_train_select_from_model,X_test_select_from_model,y_train,y_test) ## 9 features
```

## 4. What is the optimal number of features that need to be selected?

```python
def runRandomForestClassifier(X_train,X_test,y_train,y_test):

    ## Model Building
    
    RFE = RandomForestClassifier(max_depth=7, random_state=123)
    
    ## Model Training
    
    RFE_classifier = RFE.fit(X_train,y_train)
    
    ## Model Testing
    
    y_pred_train = RFE_classifier.predict(X_train)
    y_pred_test = RFE_classifier.predict(X_test)
    
    ## Model Evaluation
    
    print("Confusion Matrix:\n", confusion_matrix(y_test,y_pred_test))
    print("Accuracy Score:\t", accuracy_score(y_test,y_pred_test))
    print("Precision Score:\t", precision_score(y_test,y_pred_test))
    print("Recall Score:\t", recall_score(y_test,y_pred_test))
```

```python
for i in range(1,31):
    
    rfe_model = RFE(estimator= GradientBoostingClassifier() ,n_features_to_select=i)
    rfe_model.fit(X_train,y_train)
    
    X_train_rfe_model = rfe_model.transform(X_train)
    X_test_rfe_model = rfe_model.transform(X_test)
    print("Selected Features:", i)
    
    runRandomForestClassifier(X_train_rfe_model,X_test_rfe_model,y_train,y_test)
    print("************************")
```
